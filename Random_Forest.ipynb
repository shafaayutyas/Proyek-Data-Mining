```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import confusion_matrix, accuracy_score, classification_report
from sklearn.preprocessing import LabelEncoder

df = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/drug200.csv')
df.head()
```

1. Data Cleaning

```python
df.duplicated().sum()
```

```python
df.isnull().sum()
```

2. Data Preparation

```python
#Encoding
lebel_encoder = LabelEncoder()
df['Sex'] = lebel_encoder.fit_transform(df['Sex'])
df['BP'] = lebel_encoder.fit_transform(df['BP'])
df['Cholesterol'] = lebel_encoder.fit_transform(df['Cholesterol'])
df['Drug'] = lebel_encoder.fit_transform(df['Drug'])
df.head()
```

```python
#spliting data
x = df.drop(columns=['Drug'])
y = df['Drug']
print('X : ', x.shape)
print('Y : ', y.shape)
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

3. Modeling

```python
model = RandomForestClassifier(random_state=42)
model.fit(x_train, y_train)
y_pred = model.predict(x_test)
```

```python
RandomForest_acc = accuracy_score(y_test, y_pred)
print(classification_report(y_test, y_pred))
print('Accuracy KNN: {:.2f}%'.format(RandomForest_acc*100))
```

```python
print("Confusion matrix:")
print(confusion_matrix(y_test, y_pred))
```

4. Testing

```python
testing = {'Age': [40],
           'Sex': [1],
           'BP': [1],
           'Cholesterol': [0],
           'Na_to_K': [4.0]}

testing = pd.DataFrame(testing)
testing
```

```python
# Mengembalikan hasil prediksi angka ke string asli menggunakan inverse_transform
pred_coba = model.predict(testing)
pred_original = lebel_encoder.inverse_transform(pred_coba)
print("Hasil prediksi dari data pasien baru (String):")
print(pred_original[0])
```
